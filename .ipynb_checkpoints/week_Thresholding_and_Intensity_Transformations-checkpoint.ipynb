```python
# Thresholding_and_Intensity_Transformations
# TR: Eşikleme ve Piksel Yoğunluğu Dönüşümleri
```

```python

# Bu bölümde, gri seviyeli görüntülerin piksel değerleri üzerinde yapılan temel 
# işlemler anlatılır.
# Yani, görüntünün parlaklık ve kontrastını değiştiren, belirli piksel değerlerini dönüştüren işlemler uygulanır.
# Amaç, görüntüdeki nesneleri daha belirgin hale getirmek veya analiz için 
# hazırlanmış siyah-beyaz maskeler oluşturmaktır.
```

```python
# Bu konu başlığı altında yapılan işlemler:

    # 1. Görüntü Okuma ve Griye Çevirme
        # cv2.imread() ile görüntü dosyasını okumak
        # cv2.cvtColor() ile BGR’den griye dönüştürmek
        # Görüntü griye çevrilerek tek kanallı hale getirilir

    # 2. Farklı Thresholding Yöntemleri
        # cv2.threshold() ile belirli eşik değerlerine göre pikseller dönüştürülür
        # Kullanılan yöntemler:
            # a) Binary Thresholding
                # Eşik değerinden büyük pikseller beyaz (255), küçük pikseller siyah (0) olur
            # b) Binary Inverse Thresholding
                # Binary’nin tersi; büyük pikseller siyah, küçük pikseller beyaz olur
            # c) Trunc Thresholding
                # Eşik değerinden büyük pikseller eşik değerine eşitlenir, küçükler değişmez
            # d) ToZero Thresholding
                # Eşik değerinden küçük pikseller 0 olur, büyük pikseller aynı kalır
            # e) ToZero Inverse Thresholding
                # ToZero’nun tersi; eşik değerinden büyük pikseller 0 olur, küçükler 
                #    aynı kalır

    # 3. Görselleştirme
        # matplotlib ile tüm thresholding sonuçları bir arada gösterilir
        # plt.subplot() ile farklı yöntemler yan yana veya alt alta karşılaştırılır
        # plt.savefig() ile görselleştirme dosya olarak kaydedilir

    # 4. Görüntü Boyutlandırma (Resize)
        # Görüntü büyütme ve küçültme işlemleri yapılır
        # larger_img() fonksiyonu ile orijinal boyutun %25’i kadar büyütülür
        # smaller_img() fonksiyonu ile orijinal boyutun %25’i kadar küçültülür
        # cv2.resize() ile yeni boyutlu görüntüler elde edilir
        # Amaç, görüntüyü analiz için farklı boyutlarda kullanabilmek
```

```python
import numpy as np
import matplotlib.pyplot as plt
import cv2
```

```python
# Python da path kullanımı : 
# ./ ifadesi, kod dosyanızın çalıştığı/açıldığı klasör anlamına gelir.
# ../ ifadesi ise bir üst kasör anlamına gelir
# örneğin "c:\\users\\kullanıcı" olarak arartırsınız
# r "c://users//kullanıcı" şeklinde verebilirsiniz.
# Ancak, programlama dillerinde özellikle python da ifadesi ile o an aktif olarak 
#  çalıştığınız klasör anlamına gelir.
# her seferinde path sifirdan yazmazsınız.
```

```python
my_img = cv2.imread("../image/Lion_waiting_in_Namibia.jpeg")
```

```python
print(type(my_img))
print(my_img.shape)
```

```python
# Görüntünün yüksekliğini (satır sayısını) ekrana yazdırır
print(f'Görüntünün yüksekliği (satır sayısı) : {my_img.shape[0]}')

# Görüntünün genişliğini (sütun sayısını) ekrana yazdırır
print(f"Görüntünün genişliği (sütun sayısı) : {my_img.shape[1]}")

# Görüntünün renk kanalı sayısını (örneğin RGB için 3) ekrana yazdırır
print(f"Görüntünün renk kanalı sayısı : {my_img.shape[2]}")

# my_img.shape ifadesi, bir görüntünün boyutlarını verir.
```

```python
# BGR olarak renkli bir şekilde okunan görüntüyü, gri tonlamaya (tek kanallı) çeviriyoruz
my_img_gray = cv2.cvtColor(my_img, cv2.COLOR_BGR2GRAY)

# Gri tonlamaya çevrilen görüntünün boyutlarını yazdırıyoruz
print(my_img_gray.shape)
```

```python
# Thresholding siyah-byeaz yani gri seviyede tek kanallı görüntülerin kanal tespiti, nesne
#  tanıma, nesne ayırt etme, tıbbi alanda hücre görüntülerinin maskelenmesi (ground truth mask images)
#  gibi alanlarda kullanılan klasik bir yöntemdir.

# Bir eşik değeri seçilir, siyah-beyaz gri görüntünün her bir pikseli bu eşik değeriyle karşılaştırılır.
# Eğer ilgili piksel değeri bu eşik değerinden düşükse yeni bir piksel değeri alır, yüksekse yine yeni
#  piksel değeri alır.
# Farklı eşikleme yöntemlerinin olmasının nedeni , bu yeni piksel değerinin kaç olacağıyla ilgilidir.

# Binary Thresholding : görüntüde yer alan herhangi bir piksel seçilen eşik değerinden küçükse 
# (thresh value) siyah rengine dönüşür.
# yani 0 değerini alır, büyük veya eşitse beyaz rengine dönüşür yani 255 olur.
# Binarynin anlamı, oluşacak yeni görüntüde ya 0 ya da 255 değeri olmasıdır.
# Yani görüntüde siyah ve beyazdan başka bir renk  ve 0 ile 255 ten başka bir piksel değeri yoktur.
```

```python
retval, my_binary_thresh_img = cv2.threshold(
    my_img_gray,      # Gri tonlamalı görüntü
    thresh=125,       # Eşik değeri
    maxval=255,       # Eşiğin üstünde olan piksellerin alacağı değer
    type=cv2.THRESH_BINARY  # Eşikleme türü (Binary)
)

# cv2.threshold

# Gri tonlamalı görüntüyü ikili (binary) görüntüye çevirir.
# Piksel değerleri belirli bir eşik değerine göre 0 veya maxval olarak değiştirilir.

# retval → kullanılan eşik değeri (çoğu zaman verdiğiniz thresh ile aynı).

# my_binary_thresh_img → eşiklenmiş siyah-beyaz (binary) görüntü.
```

```python
# Eşikleme yönteminni sayısal olarak ispatı

# Belirli bir pikselin orijinal gri değeri
print(f"Orijinal Piksel Değeri : {my_img_gray[500,900]}")

# Aynı pikselin binary eşikleme sonrası değeri
print(f"Binary Thresholding Sonrası Piksel Değeri : {my_binary_thresh_img[500,900]}")
```

```python
fig = plt.figure(figsize=(10,7))  # Yeni bir figür oluştur, boyutu 10x7 inch

# Sol tarafa orijinal gri görüntüyü çiz
plt.subplot(1,2,1)                 # 1 satır, 2 sütun, 1. subplot
plt.imshow(my_img_gray, cmap='gray')  # Gri tonlama ile görüntüle
plt.axis('off')                     # Eksenleri kapat
plt.title('Original Gray Image')    # Başlık ekle

# Sağ tarafa binary eşiklenmiş görüntüyü çiz
plt.subplot(1,2,2)                 # 1 satır, 2 sütun, 2. subplot
plt.imshow(my_binary_thresh_img, cmap='gray')  # Gri tonlama ile görüntüle
plt.axis('off')                     # Eksenleri kapat
plt.title('Binary Thresh Image')    # Başlık ekle
```

```python
_, my_binary_inv_thresh_img = cv2.threshold(
    my_img_gray,        # Gri tonlamalı giriş görüntüsü
    thresh=125,         # Eşik değeri
    maxval=255,         # Eşiğin üstünde olan piksellerin alacağı değer
    type=cv2.THRESH_BINARY_INV  # Ters binary eşikleme
)

# Normal binary → nesneler beyaz, arka plan siyah

# Ters binary → nesneler siyah, arka plan beyaz
```

```python
# Orijinal gri tonlamalı görüntüdeki (500,900) konumundaki piksel değerini yazdır
print(f"Orijinal Piksel Değeri : {my_img_gray[500,900]}")

# Normal binary eşikleme sonrası aynı pikselin değerini yazdır
# Piksel ≥ 125 ise 255, < 125 ise 0 olur
print(f"Binary Thresholding Sonrası Piksel Değeri : {my_binary_thresh_img[500,900]}")

# Ters binary eşikleme sonrası aynı pikselin değerini yazdır
# Piksel ≥ 125 ise 0, < 125 ise 255 olur
print(f"Binary İnverse Thresholding Sonrası Piksel Değeri : {my_binary_inv_thresh_img[500,900]}")

# Normal binary → nesne beyaz, arka plan siyah

# Ters binary → nesne siyah, arka plan beyaz

# Bu kod, eşikleme ve ters eşikleme işleminin sayısal olarak etkisini gösteriyor.
```

```python
# Yeni bir figür oluşturuyoruz, boyutu 10x8 inch
fig = plt.figure(figsize=(10,8))

# 2 satır, 2 sütunluk bir düzenin 1. hücresine orijinal gri görüntüyü yerleştiriyoruz
plt.subplot(2,2,1)
plt.imshow(my_img_gray, cmap='gray')  # Gri tonlamayla görüntüle
plt.axis('off')                        # Eksenleri gizle
plt.title('Original Gray Image')       # Başlık ekle

# 2 satır, 2 sütunluk düzenin 2. hücresine normal binary eşiklenmiş görüntüyü yerleştiriyoruz
plt.subplot(2,2,2)
plt.imshow(my_binary_thresh_img, cmap='gray')  # Binary görüntüyü gri tonlamayla göster
plt.axis('off')                                # Eksenleri gizle
plt.title('Binary Thresh Image')              # Başlık ekle

# 2 satır, 2 sütunluk düzenin 3. hücresine ters binary eşiklenmiş görüntüyü yerleştiriyoruz
plt.subplot(2,2,3)
plt.imshow(my_binary_inv_thresh_img, cmap='gray')  # Ters binary görüntüyü gri tonlamayla göster
plt.axis('off')                                     # Eksenleri gizle
plt.title('Binary Inverse Thresh Image')          # Başlık ekle
```

```python
# TRUNC THRESHOLDING ilgili piksel değeri eşik değerinden büyükse, eşik değerine eşitler,
# eşik değerinden küçükse olduğu gibi bırakır.
# Kullanım amacı ; görüntüdeki yüksek değere sahip piksellerin devre dışı bırakılmasıdır/kırpılmasıdır.

# Trunc (kesme) eşikleme yöntemi ile gri tonlamalı görüntüyü eşikliyoruz
# cv2.threshold fonksiyonu iki değer döndürür:
# 1) retval : kullanılan eşik değeri (çoğu zaman verdiğimiz thresh ile aynı)
# 2) eşiklenmiş görüntü
# Burada retval'i kullanmayacağımız için '_' (underscore) ile görmezden geliyoruz
_, my_trunc_thresh_img = cv2.threshold(
    my_img_gray,      # Gri tonlamalı giriş görüntüsü
    thresh=125,       # Eşik değeri
    maxval=255,       # Trunc için aslında kullanılmaz ama parametre olarak gerekli
    type=cv2.THRESH_TRUNC  # Trunc eşikleme tipi
)

# Sonuç: my_trunc_thresh_img → piksel değeri eşik değerinden büyük olanlar eşik değeri ile sınırlanır,
#          küçük olanlar olduğu gibi kalır.

# _ → “Bu değeri alıyorum ama kullanmayacağım.”

# Trunc threshold → parlak pikseller kesilir, koyu pikseller korunur.
```

```python
# Orijinal gri tonlamalı görüntüdeki (500,900) konumundaki piksel değeri
print(f"Orijinal Piksel Değeri : {my_img_gray[500,900]}")

# Normal binary eşikleme sonrası aynı pikselin değeri
# Eğer piksel ≥ 125 → 255, < 125 → 0
print(f"Binary Thresholding Sonrası  Değeri : {my_binary_thresh_img[500,900]}")

# Ters binary eşikleme sonrası aynı pikselin değeri
# Eğer piksel ≥ 125 → 0, < 125 → 255
print(f"Binary İnverse Thresholding  Değeri : {my_binary_inv_thresh_img[500,900]}")

# Trunc (kesme) eşikleme sonrası aynı pikselin değeri
# Eğer piksel ≥ 125 → 125, < 125 → piksel değeri değişmez
print(f"Trunc Thresholding  Değeri : {my_trunc_thresh_img[500,900]}")


# Aynı piksel için farklı threshold yöntemlerinin etkisini sayısal olarak gösterir.

# Binary, Inverse Binary ve Trunc yöntemlerinin mantığını tek bir bakışta görebilirsin.
```

```python
# Yeni bir matplotlib figürü oluşturuyoruz, boyutu 10x7 inch
fig = plt.figure(figsize=(10,7))

# 2 satır × 2 sütun düzeninde 1. subplot: Orijinal gri görüntü
plt.subplot(2,2,1)
plt.imshow(my_img_gray, cmap='gray')  # Gri tonlamayla göster
plt.axis('off')                        # Eksenleri kapat
plt.title('Original Gray Image')       # Başlık ekle

# 2. subplot: Normal binary eşiklenmiş görüntü
plt.subplot(2,2,2)
plt.imshow(my_binary_thresh_img, cmap='gray')  # Binary görüntüyü gri tonlamayla göster
plt.axis('off')                                # Eksenleri kapat
plt.title('Binary Thresh Image')              # Başlık ekle

# 3. subplot: Ters binary eşiklenmiş görüntü
plt.subplot(2,2,3)
plt.imshow(my_binary_inv_thresh_img, cmap='gray')  # Inverse binary görüntüyü göster
plt.axis('off')                                     # Eksenleri kapat
plt.title('Binary Inverse Thresh Image')          # Başlık ekle

# 4. subplot: Trunc (kesme) eşiklenmiş görüntü
plt.subplot(2,2,4)
plt.imshow(my_trunc_thresh_img, cmap='gray')  # Trunc eşikleme sonucu görüntü
plt.axis('off')                                # Eksenleri kapat
plt.title('TRUNC Thresh Image')               # Başlık ekle
```

```python
# TOZERO THRESHOLDING ilgili piksel değeri eşik değerinden büyükse, olduğu gibi bırakır, eşik değerinden küçükse 0 yapar.
# kullanım amacı, görüntüdeki düşük değere sahip piksellerin devre dışı bırakmaktır


# TOZERO (sıfırın altına sıfır koyma) eşikleme yöntemi ile gri tonlamalı görüntüyü eşikliyoruz
# cv2.threshold fonksiyonu iki değer döndürür:
# 1) retval : kullanılan eşik değeri (çoğu zaman verdiğimiz thresh ile aynı)
# 2) eşiklenmiş görüntü
# Burada retval'i kullanmayacağımız için '_' (underscore) ile görmezden geliyoruz
_, my_tozero_img = cv2.threshold(
    my_img_gray,       # Gri tonlamalı giriş görüntüsü
    thresh=125,        # Eşik değeri
    maxval=255,        # TOZERO için aslında kullanılmaz ama parametre olarak gerekli
    type=cv2.THRESH_TOZERO  # TOZERO eşikleme tipi
)

# Sonuç: my_tozero_img → Piksel değeri eşik değerinden küçük olanlar 0 yapılır,
#          eşik değerinden büyük olanlar ise orijinal değerini korur.
```

```python
# Orijinal gri tonlamalı görüntüdeki (500,900) konumundaki piksel değeri
print(f"Orijinal Piksel Değeri : {my_img_gray[500,900]}")

# Normal binary eşikleme sonrası aynı pikselin değeri
# Eğer piksel ≥ 125 → 255, < 125 → 0
print(f"Binary Thresholding Sonrası  Değeri : {my_binary_thresh_img[500,900]}")

# Ters binary eşikleme sonrası aynı pikselin değeri
# Eğer piksel ≥ 125 → 0, < 125 → 255
print(f"Binary İnverse Thresholding  Değeri : {my_binary_inv_thresh_img[500,900]}")

# Trunc (kesme) eşikleme sonrası aynı pikselin değeri
# Eğer piksel ≥ 125 → 125, < 125 → piksel değeri değişmez
print(f"Trunc Thresholding  Değeri : {my_trunc_thresh_img[500,900]}")

# TOZERO eşikleme sonrası aynı pikselin değeri
# Eğer piksel ≥ 125 → orijinal değer korunur, < 125 → 0 olur
print(f"TOZERO  Değeri : {my_tozero_img[500,900]}")
```

```python
# Yeni bir matplotlib figürü oluşturuyoruz, boyutu 10x7 inch
fig = plt.figure(figsize=(10,7))

# 3 satır × 2 sütun düzeninde 1. subplot: Orijinal gri görüntü
plt.subplot(3,2,1)
plt.imshow(my_img_gray, cmap='gray')  # Gri tonlamayla göster
plt.axis('off')                        # Eksenleri kapat
plt.title('Original Gray Image')       # Başlık ekle

# 2. subplot: Normal binary eşiklenmiş görüntü
plt.subplot(3,2,2)
plt.imshow(my_binary_thresh_img, cmap='gray')  # Binary görüntüyü gri tonlamayla göster
plt.axis('off')                                # Eksenleri kapat
plt.title('Binary Thresh Image')              # Başlık ekle

# 3. subplot: Ters binary eşiklenmiş görüntü
plt.subplot(3,2,3)
plt.imshow(my_binary_inv_thresh_img, cmap='gray')  # Inverse binary görüntüyü göster
plt.axis('off')                                     # Eksenleri kapat
plt.title('Binary Inverse Thresh Image')          # Başlık ekle

# 4. subplot: Trunc (kesme) eşiklenmiş görüntü
plt.subplot(3,2,4)
plt.imshow(my_trunc_thresh_img, cmap='gray')  # Trunc eşikleme sonucu görüntü
plt.axis('off')                                # Eksenleri kapat
plt.title('TRUNC Thresh Image')               # Başlık ekle

# 5. subplot: TOZERO eşiklenmiş görüntü
plt.subplot(3,2,5)
plt.imshow(my_tozero_img, cmap='gray')  # TOZERO threshold sonucu görüntü
plt.axis('off')                          # Eksenleri kapat
plt.title('TOZERO Thresh Image')         # Başlık ekle
```

```python
_, my_tozero_inv_img = cv2.threshold(my_img_gray,thresh=125,maxval=255,type=cv2.THRESH_TOZERO_INV)
```

```python
# Orijinal gri tonlamalı görüntüde (my_img_gray) belirli bir pikselin (500, 900) konumundaki değeri yazdırır.
# Gri tonlamalı (grayscale) görüntülerde piksel değeri 0-255 arasıdır.
# 0 → siyah, 255 → beyaz, aradaki değerler gri tonlarını temsil eder.
print(f"Orijinal Piksel Değeri : {my_img_gray[500,900]}")

# Binary (ikili) thresholding sonrası aynı pikselin değerini yazdırır.
# Binary thresholding: piksel değeri belirli bir eşikten büyükse 255 (beyaz), küçükse 0 (siyah) yapılır.
# Bu, görüntüyü tamamen siyah-beyaz hale getirir.
print(f"Binary Thresholding Sonrası  Değeri : {my_binary_thresh_img[500,900]}")

# Binary Inverse thresholding sonrası aynı pikselin değerini yazdırır.
# Binary inverse: normal binary’nin tersi şekilde çalışır.
# Eşik değeri üzerindekiler 0 (siyah), altındakiler 255 (beyaz) yapılır.
print(f"Binary İnverse Thresholding  Değeri : {my_binary_inv_thresh_img[500,900]}")

# Truncate (kesme) thresholding sonrası piksel değeri.
# Trunc thresholding: piksel değeri eşikten büyükse eşik değerine eşitlenir, küçükse olduğu gibi kalır.
# Yani maksimum parlaklık değeri sınırlandırılır.
print(f"Trunc Thresholding  Değeri : {my_trunc_thresh_img[500,900]}")

# ToZero thresholding sonrası piksel değeri.
# ToZero: piksel değeri eşikten küçükse 0 yapılır (yani karartılır), büyükse olduğu gibi kalır.
# Bu yöntem karanlık bölgeleri sıfırlar ama parlak bölgeleri korur.
print(f"Tozero  Değeri : {my_tozero_img[500,900]}")

# ToZero Inverse thresholding sonrası piksel değeri.
# ToZero inverse: normal ToZero’nun tersidir.
# Piksel değeri eşikten büyükse 0 yapılır, küçükse olduğu gibi kalır.
# Bu da parlak bölgeleri karartır, karanlık bölgeleri korur.
print(f"Tozero INV Değeri : {my_tozero_inv_img[500,900]}")
```

```python
# Resize: satır x sütun bazında büyütme/küçültme işlemi

#cv2 görüntüyü resize ederken kullanılan formül

# cv2.resize(original_image,(new_width,new_height),cv2.INTER_LANCZOS4)

# 2 adet resize formülü, bu formül tek parametre alıcak o da görüntüyü saklayacak
# bir fonk büyütmek için öbürü ise küçültme için
# büyütme ve küçültme için de olsa görüntünün yükseklil ve genişliğinin %25 oranında büyütecek ya da küçültücek
# bu yeni görüntüyü de geriye döndürcek, siz de bunu çalıştırırken bir değişkene atıcaksınız.
# önce .shape ile orijinal görüntüyü ve büyük/küçük görüntüyü boyut olarak kontrol ediceksiniz.

# GÖRÜNTÜYÜ YENİDEN BOYUTLANDIRMA (Resize)
# Bu işlem, görüntünün satır (yükseklik) ve sütun (genişlik) boyutlarını değiştirmek için kullanılır.
# OpenCV'de cv2.resize() fonksiyonu ile yapılır.

# Genel kullanım:
# cv2.resize(original_image, (new_width, new_height), interpolation_method)
# interpolation_method -> cv2.INTER_LANCZOS4 yüksek kaliteli ölçekleme sağlar (büyütmede daha net sonuç verir).


# GÖRÜNTÜYÜ %25 ORANINDA KÜÇÜLTEN FONKSİYON
def smaller_img(image):
    # image.shape -> (yükseklik, genişlik, kanal_sayısı)
    # Burada genişlik (shape[1]) ve yükseklik (shape[0]) alınır.

    # Yeni genişlik: mevcut genişliğin %75'i (%25 küçültme)
    width  = int(image.shape[1] - int(image.shape[1] * 0.25))

    # Yeni yükseklik: mevcut yüksekliğin %75'i (%25 küçültme)
    height = int(image.shape[0] - int(image.shape[0] * 0.25))

    # Görüntüyü yeniden boyutlandır (interpolasyon yöntemi olarak LANCZOS4 kullan)
    result = cv2.resize(image, (width, height), cv2.INTER_LANCZOS4)

    # Yeni boyutu terminale yazdır (yükseklik, genişlik, kanal sayısı)
    print(f"Küçültülmüş Boyutu: {result.shape}")

    # Küçültülmüş görüntüyü geri döndür
    return result

# GÖRÜNTÜYÜ %25 ORANINDA BÜYÜTEN FONKSİYON

def larger_img(image):
    # Yeni genişlik: mevcut genişliğin %125'i (%25 büyütme)
    width  = int(image.shape[1] + int(image.shape[1] * 0.25))

    # Yeni yükseklik: mevcut yüksekliğin %125'i (%25 büyütme)
    height = int(image.shape[0] + int(image.shape[0] * 0.25))

    # Görüntüyü yeniden boyutlandır
    result = cv2.resize(image, (width, height), cv2.INTER_LANCZOS4)

    # Yeni boyutu terminale yazdır
    print(f"Büyütülmüş Boyutu: {result.shape}")

    # Büyütülmüş görüntüyü geri döndür
    return result
```

```python
# Görüntüyü %25 oranında büyüt ve sonucu my_larger_img değişkenine ata
my_larger_img = larger_img(my_img_gray)

# Görüntüyü %25 oranında küçült ve sonucu my_smaller_img değişkenine ata
my_smaller_img = smaller_img(my_img_gray)

# Orijinal görüntünün boyutunu yazdır
# .shape -> (yükseklik, genişlik, kanal_sayısı)
# Gri görüntüde genelde 2 boyut olur (yükseklik, genişlik)
print(f"Orijinal  Boyutu : {my_img_gray.shape}")

# Büyütülmüş görüntünün boyutunu yazdır
print(f"Büyütülmüş  Boyutu : {my_larger_img.shape}")

# Küçültülmüş görüntünün boyutunu yazdır
print(f"Küçültülmüş  Boyutu : {my_smaller_img.shape}")
```
