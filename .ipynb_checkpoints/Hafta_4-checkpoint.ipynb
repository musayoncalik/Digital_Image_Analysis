```python
import numpy as np
import matplotlib.pyplot as plt
import cv2
import os
```

```python
# Morfolojik İşlemler : siyah-beyaz(binary) görüntüler üzerinde, 
# görüntünün yapısal özelliklerinin değiştirilmesine yarayan işlemler bütünüdür.

# 4 temel morfolojik işlem vardır .
# 
```

```python

my_binary_img = cv2.imread("./deneme.png", 0)

print(my_binary_img.shape)
```

```python
my_kernel = np.ones((5,5), dtype = np.uint8)
```

```python
my_erosion_img = cv2.erode(my_binary_img, my_kernel, iterations = 2)
```

```python
my_dilation_img = cv2.dilate(my_binary_img, my_kernel, iterations = 6)
```

```python
import matplotlib.pyplot as plt

fig = plt.figure(figsize=(10,7))

# 1. Görsel: Orijinal
plt.subplot(1,3,1)
plt.imshow(my_binary_img, cmap="gray")  #  plt.show() yerine plt.imshow()
plt.axis("off")                         #  plt.axis = (...) değil, plt.axis("off")
plt.title("Original Image")

# 2. Görsel: Erosion
plt.subplot(1,3,2)
plt.imshow(my_erosion_img, cmap="gray") #  'cmaps' değil 'cmap'
plt.axis("off")
plt.title("Erosion Image")


# 2. Görsel: Erosion
plt.subplot(1,3,3)
plt.imshow(my_dilation_img, cmap="gray") #  'cmaps' değil 'cmap'
plt.axis("off")
plt.title("Dilate Image")

plt.show()  
```

```python
new_img = cv2.imread("./deneme1.png", 0)
```

```python
new_img.shape
```

```python
white_noise = np.random.randint(0,2,size = new_img.shape[:2])
white_noise *=255
```

```python
plt.imshow(white_noise, cmap="gray")
plt.axis("off")
plt.title("Full Noised Image")
```

```python
my_noised_img = white_noise + new_img
plt.imshow(my_noised_img, cmap="gray")
plt.axis("off")
plt.title("Noised Image")
```

```python
# Opening : Görüntüye, önce erosian sonra da dilation işlemini arka arkaya uygular.
# Önce, görüntüdeki
```

```python
my_opnening_img = cv2.morphologyEx(my_noised_img.astype(np.float32),
                                  cv2.MORPH_OPEN, my_kernel)
plt.imshow(my_opnening_img, cmap="gray")
plt.axis("off")
plt.title("Opening Image")
                                    
```

```python
#################################
plt.figure(figsize=(16, 4))

# 1. Görsel: Full Noised Image
plt.subplot(1, 4, 1)
plt.imshow(new_img, cmap="gray")
plt.axis("off")
plt.title("Full Noised Image")

# 2. Görsel: Noised Image
plt.subplot(1, 4, 2)
plt.imshow(my_noised_img, cmap="gray")
plt.axis("off")
plt.title("Noised Image")

# 3. Görsel: Erosion
plt.subplot(1, 4, 3)
plt.imshow(my_erosion_img, cmap="gray")
plt.axis("off")
plt.title("Erosion Image")

# 4. Görsel: Dilation
plt.subplot(1, 4, 4)
plt.imshow(new_img, cmap="gray")
plt.axis("off")
plt.title("Dilation Image")

plt.tight_layout()
plt.show()
```

```python
# Closing (kapanma) : 
```

```python
#blac_noise = np.random.randint(0,2, size=new_img.shape[0:2])
#black_noise *= -255

# Black noise oluşturma
black_noise = np.random.randint(0, 2, size=new_img.shape[0:2]).astype(np.float32)  # float32 yapıyoruz
black_noise *= -255  # artık sorunsuz çarpılır
```

```python
my_black_noise_img = black_noise + new_img
```

```python
my_black_noise_img[my_black_noise_img <= -245] = 0

plt.imshow(my_black_noise_img, cmap = "gray")
plt.axis("off")
plt.title("Black Noised Image")
```

```python
my_closing_img = cv2.morphologyEx(my_black_noise_img.astype(np.float32),
                                  cv2.MORPH_OPEN, my_kernel)
plt.imshow(my_black_noise_img, cmap="gray")
plt.axis("off")
plt.title("Closing Image")
```

```python
plt.figure(figsize=(16, 4))

# 1. Görsel: Original Image
plt.subplot(1, 4, 1)
plt.imshow(new_img, cmap="gray")
plt.axis("off")
plt.title("Original Image")

# 2. Görsel: Noised Image (black noise eklenmiş)
plt.subplot(1, 4, 2)
plt.imshow(my_black_noise_img, cmap="gray")
plt.axis("off")
plt.title("Noised Image")

# 3. Görsel: Erosion
plt.subplot(1, 4, 3)
plt.imshow(my_erosion_img, cmap="gray")  # Closing yerine Erosion kullanılmalı
plt.axis("off")
plt.title("Erosion Image")

# 4. Görsel: Dilation
plt.subplot(1, 4, 4)
plt.imshow(my_dilation_img, cmap="gray")  # Original image yerine Dilation kullanılmalı
plt.axis("off")
plt.title("Dilation Image")

plt.tight_layout()
plt.show()
```

```python
my_gradient_1 = cv2.morphologyEx(my_binary_img, cv2.MORPH_GRADIENT, my_kernel)
my_gradient_2 = cv2.morphologyEx(new_img, cv2.MORPH_GRADIENT, my_kernel)
```

```python
fig = plt.figure(figsize=(10,7))


# 
plt.subplot(2, 2, 1)
plt.imshow(my_binary_img, cmap="gray")
plt.axis("off")
plt.title("Original Image")

# 
plt.subplot(2, 2, 2)
plt.imshow(my_gradient_1, cmap="gray")
plt.axis("off")
plt.title("Gradient 1  Image")

plt.subplot(2, 2, 3)
plt.imshow(new_img, cmap="gray")
plt.axis("off")
plt.title("Original Image")

plt.subplot(2, 2, 4)
plt.imshow(my_gradient_2, cmap="gray")
plt.axis("off")
plt.title("Gradient 2 Image")

plt.tight_layout()
plt.show()
```
