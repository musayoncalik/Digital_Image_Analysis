```python
import cv2
import matplotlib.pyplot as plt
import numpy as np
```

```python
# Görüntüyü, renkli olsa bile her zaman gri seviyede okumalıyız.
# ya BGR olarak okuyup sonradan gray seviyesine çevirirsiniz.
# ya da direkt başlangıçta gray seviyede okursunuz.
# NOT : hiç bir işlem uygulanmadığında görüntü renklidir.

my_bgr_img = cv2.imread("./Hafta_6/bonus.jpg")
my_img = cv2.imread("./Hafta_6/bonus.jpg", 0)

# Daha sonra görüntüye eşikleme işlemi uygularız. 
# Eşikleme işlemiyle görüntünün zaten siyah-beyaz seviyede ayrımı ortaya çıkar.

ret, thresh = cv2.threshold(my_img, 127, 255, cv2.THRESH_BINARY)

# Görüntüdeki kontuları bulma

countur, hierarchy = cv2.findContours(thresh, cv2.RETR_EXTERNAL, cv2.CHAIN_APPROX_SIMPLE)

# yukarıdaki kontur bulma fonksiyonundan 2 değer döner.
# İlk değer, bulunan konturların listesini içerir.
# İkinci değer bu konturlar arasındaki ebeveyn ilişkisini gösterir.
# Bize lazım olan, göstereceğimiz asıl değer olan konturların listesini saklayan ilk değişkendir.
# Elde ettiğimiz konturları görüntü üzerinde birleştirip çizdiriyoruz.

cv2.drawContours(my_bgr_img, countur, -1, (0,255,0),2)

# Fonksiyondaki parametreler: 
# -1 : bütün konturları göster, satır-sütun tamamı olacak şekilde 
# (0,255,0) RGB nin kodlarıdır. Bu renk kodu yeşili temsil eder.
# Konturu çizdirirken yeşil renkle çizdirecek
# 2 değeri de çizginin (yeşil renkle çizdirilecek kontur çizgisinin ) kalınlığıdır.

cv2.imshow("Original Countured Image", my_bgr_img)
cv2.waitKey(0)
cv2.destroyWindow()
```

```python
import cv2

# Görüntüyü, renkli olsa bile gri seviyede okumalıyız
# ya BGR olarak okuyup sonradan gray seviyesine çevirirsiniz
# ya da direkt başlangıçta gray seviyede okursunuz
# NOT: Hiçbir işlem uygulanmadığında görüntü renklidir

# Hem renkli hem gri görüntüyü aynı dosyadan okuyalım
my_bgr_img = cv2.imread("./Hafta_6/deneme.jpg")  # Renkli
my_img = cv2.imread("./Hafta_6/deneme.jpg", 0)   # Gri seviye

# Eşikleme (threshold) ile siyah-beyaz seviyelere ayırma
ret, thresh = cv2.threshold(my_img, 127, 255, cv2.THRESH_BINARY)

# Görüntüdeki konturları bulma
contours, hierarchy = cv2.findContours(thresh, cv2.RETR_EXTERNAL, cv2.CHAIN_APPROX_SIMPLE)
# contours → konturların listesi
# hierarchy → konturlar arası ebeveyn-çocuk ilişkisi (biz contours kullanacağız)

# Konturları görüntü üzerinde çizdirme
cv2.drawContours(my_bgr_img, contours, -1, (0,255,0), 2)
# Parametreler:
# - -1 → tüm konturlar çizilsin
# - (0,255,0) → yeşil renk kodu (RGB)
# - 2 → çizgi kalınlığı

# Görüntüyü gösterme
cv2.imshow("Original Countoured Image", my_bgr_img)
cv2.waitKey(0)
cv2.destroyWindows()
```
