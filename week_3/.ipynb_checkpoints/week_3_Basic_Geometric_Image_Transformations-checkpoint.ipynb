```python
# BASIC GEOMETRIC IMAGE TRANSFORMATIONS (Temel Geometrik Görüntü İşlemleri)
```

```python
# Bu bölümde, görüntünün geometrik özellikleri üzerinde yapılan temel işlemler anlatılır.
# Yani, görüntünün konumunu veya şeklini değiştiren ama içeriğini bozmayan dönüşümler uygulanır.
# Amaç, görüntüyü farklı açılardan görmek veya belirli bir bölgeyi ayırmaktır.
```

```python
# Bu konu başlığı altında yapılan işlemler:
	# 1. Görüntü Okuma ve Görselleştirme
		# cv2.imread() ile görüntü dosyasını okumak
		# cv2.cvtColor() ile BGR’den RGB’ye dönüştürmek
		# matplotlib ile görüntüyü göstermek
	# 2. Görüntü Aynalama (Flipping)
		# cv2.flip() veya [::-1] ile
		# Dikey, yatay veya hem dikey hem yatay çevirme işlemleri yapılır.
		# Görüntü ayna simetrisiyle ters çevrilir.
	# 3. Manuel Flip Fonksiyonu Yazma
		# my_flip() fonksiyonu ile cv2.flip’in mantığı öğretilir.
		# Parametreye göre (0, 1, -1) farklı eksenlerde çevirme yapılır.
	# 4. Görüntü Boyutlarını Bulma (Shape)
		#  img.shape ile yükseklik, genişlik ve kanal sayısı elde edilir.
	# 5. Görüntü Kırpma (Cropping)
		# Görüntünün belirli bir bölgesi (örneğin son %25’lik kısmı) alınır.
		# Bu işlem, görüntünün istenmeyen kısımlarını kesmek veya belirli alanları 
        # analiz etmek için kullanılır.
```

```python
# Bu konu, dijital görüntü işleme sürecinde görüntüyü çevirmek, tersine almak ve belli 
# bir kısmını kesmek gibi temel geometrik dönüşümleri öğretir.

# Yani görüntü üzerinde ayna, kesme ve konum değiştirme işlemleri yapılır. 
```

```python
import cv2          # cv2: OpenCV → görüntü okuma, yazma, dönüştürme
import numpy as np   # numpy: sayısal işlemler (piksel değerleriyle oynamak) 
import matplotlib.pyplot as plt # matplotlib: görselleştirme (resimleri göstermek için)
```

```python
# Görüntü Okuma ve Görselleştirme
my_img = cv2.imread("./Lion_waiting_in_Namibia.jpeg")  # 2 nokta (../) demek geri 
```

```python
print(type(my_img))
```

```python
# BGR’den RGB’ye renk dönüşümü yapar

my_rgb_img = cv2.cvtColor(my_img, cv2.COLOR_BGR2RGB)
```

```python
plt.imshow(my_rgb_img)  # RGB resmini göster
plt.axis("off")          # Eksenleri gizle, sadece resmi göster
```

```python
# Görüntüyü aynalama (flip)
# Görüntüyü farklı eksenlerde ters çevirir (ayna görüntüsü oluşturur).
# Flip Fonksiyonu : Görüntüyü 180 derece ters çevirir.
# Dikey eksende, yatay eksende ve hem dikey hem de yatay eksende ters çevirir.

# kullanımı : cv2.flip(orijinal_görüntü, tam_sayi_degeri)
# tam_sayi_degeri : 0 , 1, -1
# 0: dikey eksen , 1: yatay eksen, -1: hem dikey hem yatay



# Dikey eksende çevirme (0)
my_vertical_img = cv2.flip(my_rgb_img, 0)

# Yatay eksende çevirme (1)
my_horizontal_img = cv2.flip(my_rgb_img, 1)

# Hem dikey hem yatay (180 derece) (-1)
my_both_flipped_img = cv2.flip(my_rgb_img, -1)
```

```python
# Görselleri çizdir

# Yeni bir çizim alanı (figure) oluşturur ve boyutunu belirler.
fig = plt.figure(figsize=(10,7))

# plt.subplot(2,2,n)
# 2x2’lik bir grid oluşturur ve n hangi hücreye çizim yapılacağını belirtir.
# Yani aynı ekranda 4 farklı görsel yan yana veya alt alta gösterilebilir

plt.subplot(2,2,1)
plt.imshow(my_rgb_img)   # Görüntüyü gösterir. Renkli veya gri tonlamalı olabilir.
plt.axis("off")          # Eksenleri gizler; sadece görüntü görünür.
plt.title("Original Image")  # Her görselin başlığını verir

plt.subplot(2,2,2)
plt.imshow(my_vertical_img)
plt.axis("off")
plt.title("Vertical Flip")

plt.subplot(2,2,3)
plt.imshow(my_horizontal_img)
plt.axis("off")
plt.title("Horizontal Flip")

plt.subplot(2,2,4)
plt.imshow(my_both_flipped_img)
plt.axis("off")
plt.title("Both Flip (180°) || Vertical and Horizontal Image ")

plt.show()           # Tüm subplotları ekranda görüntüler.
```

```python
# Kendi flip fonksiyonunuzu yazın
# 0, 1 veya -1 parametresine göre görüntüyü ters çeviren özel bir fonksiyon tanımlanmıştır.

def my_flip(org_img, number):
    if number == 0:       # Dikey çevirme (alt-üst)
        org_img = org_img[::-1, :, :]
    elif number == 1:     # Yatay çevirme (sağ-sol)
        org_img = org_img[:, ::-1, :]
    elif number == -1:    # Hem yatay hem dikey
        org_img = org_img[::-1, ::-1, :]
    else:
        raise ValueError("Lütfen doğru parametreleri girdiğinizden emin olun! (0, 1 veya -1)")
    return org_img
```

```python
# Test
my_new_vertical = my_flip(my_rgb_img, 0)
my_new_horizontal = my_flip(my_rgb_img, 1)
my_new_both = my_flip(my_rgb_img, -1)
```

```python
# Kendi flip fonksiyonumuz
def my_flip(img, mode="vertical"):
    """
    Resmi verilen moda göre çevirir.
    
    Parametreler:
    img : numpy array, görüntü
    mode : str, çevirme modu. Seçenekler:
           'vertical'   -> dikey çevirme
           'horizontal' -> yatay çevirme
           'both'       -> hem dikey hem yatay çevirme
           'none'       -> orijinal resmi döndürme
           
    Return:
    Çevrilmiş resim
    """
    if mode == "vertical":
        img = img[::-1, :, :]       # Satırları ters çevir (dikey)
    elif mode == "horizontal":
        img = img[:, ::-1, :]       # Sütunları ters çevir (yatay)
    elif mode == "both":
        img = img[::-1, ::-1, :]   # Hem satır hem sütunları ters çevir
    elif mode == "none":
        img = img                   # Orijinal resmi döndür
    else:
        raise ValueError("Geçersiz mode! Seçenekler: 'vertical', 'horizontal', 'both', 'none'.")
    
    return img
```

```python
# Örnek kullanım: Flip işlemleri

# 1. Dikey flip
my_flip_vertical = my_flip(my_rgb_img, mode="vertical")  
# my_rgb_img resmini dikey olarak çevirir

# 2. Yatay flip
my_flip_horizontal = my_flip(my_rgb_img, mode="horizontal")  
# my_rgb_img resmini yatay olarak çevirir

# 3. Hem dikey hem yatay flip
my_flip_both = my_flip(my_rgb_img, mode="both")  
# my_rgb_img resmini hem dikey hem yatay çevirir
```

```python
# Görselleri gösterelim
plt.figure(figsize=(10,5))
plt.subplot(2,2,1); plt.imshow(my_rgb_img); plt.title("Orijinal",); plt.axis("off")
plt.subplot(2,2,2); plt.imshow(my_flip_vertical); plt.title("Dikey"); plt.axis("off")
plt.subplot(2,2,3); plt.imshow(my_flip_horizontal); plt.title("Yatay"); plt.axis("off")
plt.subplot(2,2,4); plt.imshow(my_flip_both); plt.title("Both"); plt.axis("off")
plt.show()
```

```python
# Görüntü boyutlarını öğrenmek
# Görüntünün yüksekliği, genişliği ve renk kanalı sayısı bulunur.

print(f"görüntünün yüksekliği = {my_rgb_img.shape[0]}")
print(f" görüntünün genişliği = {my_rgb_img.shape[1]}")
```

```python
# Görüntünün hem yükseklik hem genişlik açısından son %25'lik dilimini
# göstermek için hem satır hem de sütun sayısının %75'inden başlayarak
# indeksleme yaparız.
# Bu işlemin görüntü işlemedeki karşılığı CROPPING olarak geçer
```

```python
# Görüntüyü kırpmak (cropping)
# Görüntünün son %25’lik kısmı alınır (istenen bölge kesilerek çıkarılır).
yukseklik_dilim = int(my_rgb_img.shape[0] * 0.75)
genislik_dilim = int(my_rgb_img.shape[1] * 0.75)

print(f"%75 lik kısım (yükseklik) = {yukseklik_dilim}")
print(f"%75 lik kısım (genişlik) = {genislik_dilim}")
```

```python
# Görüntüyü kırp
my_crop_img = my_rgb_img[yukseklik_dilim:, genislik_dilim:, :]
```

```python
from matplotlib.patches import Rectangle # Matplotlib'den dikdörtgen çizimi için gerekli sınıfı içe aktar
from matplotlib.image import imread  # Resmi okumak için (imread fonksiyonu)


#  Çerçeve ekleme fonksiyonu
def add_frame(img, frame_color=(255,0,0), frame_width=10):
    # Bir görüntüye (img) çerçeve ekleyen fonksiyon tanımlanır.
    # frame_color varsayılanı kırmızıdır, frame_width varsayılanı 10 pikseldir.
    if frame_width == 0:
        # Çerçeve kalınlığı 0 ise, görüntünün bir kopyasını döndür ve çık.
        return img.copy()
    h, w, c = img.shape
    # Görüntünün yüksekliğini (h), genişliğini (w) ve kanal sayısını (c) al.
    
    # Çerçeveli yeni görüntü için yer aç. Boyutlar her kenardan 2*frame_width artırılır.
    # np.ones, çerçeve rengiyle doldurulacak bir dizi oluşturmak için kullanılır.
    framed_img = np.ones((h + 2*frame_width, w + 2*frame_width, c), dtype=np.uint8)
    
    framed_img[:] = frame_color
    # Yeni görüntünün tüm piksellerini çerçeve rengiyle (varsayılan: kırmızı) doldur.
    
    # Orijinal görüntüyü, çerçevenin başladığı noktaya (frame_width içeride) kopyala.
    framed_img[frame_width:frame_width+h, frame_width:frame_width+w, :] = img
    return framed_img
    # Çerçeveli görüntüyü döndür.

# Kırpılmış resme çerçeve ekle
# Tanımlanmış 'my_crop_img' değişkenine çerçeve eklenir.
# Sonuç, 'framed_crop' değişkenine atanır.
framed_crop = add_frame(my_crop_img, frame_color=(255,0,0), frame_width=10)

#  Görselleştirme
plt.figure(figsize=(12,5))
# 12 inç genişliğinde ve 5 inç yüksekliğinde yeni bir şekil (görselleştirme penceresi) oluştur.

# Orijinal resim + kırpılmış alan dikdörtgen ile
ax1 = plt.subplot(1,2,1)
# 1 satır, 2 sütunlu alt grafik düzeninin 1. alt grafiğini seç.
ax1.imshow(my_rgb_img)
# 'my_rgb_img' (orijinal resim) değişkenini bu alt grafikte göster.
ax1.set_title("Original Resim")
# Alt grafiğin başlığını ayarla.
ax1.axis("off")
# Eksenleri (x ve y koordinatları) gizle.

# Kırpılmış alanı dikdörtgen ile işaretle
# Dikdörtgen nesnesi oluşturulur.
# Başlangıç noktası: (genislik_dilim, yukseklik_dilim)
# Boyutlar: (my_rgb_img.shape[1]-genislik_dilim, my_rgb_img.shape[0]-yukseklik_dilim)
rect = Rectangle((genislik_dilim, yukseklik_dilim),
                  my_rgb_img.shape[1]-genislik_dilim,
                  my_rgb_img.shape[0]-yukseklik_dilim,
                  linewidth=3, edgecolor='red', facecolor='none')
# Çizgi kalınlığını 3, rengini kırmızı yap ve içini boş bırak.
ax1.add_patch(rect)
# Oluşturulan dikdörtgeni (kırpma alanını gösteren) ilk alt grafiğe ekle.

# Kırpılmış resim + çerçeve
ax2 = plt.subplot(1,2,2)
# 1 satır, 2 sütunlu alt grafik düzeninin 2. alt grafiğini seç.
ax2.imshow(framed_crop)
# Çerçevelenmiş kırpılmış resmi ('framed_crop') bu alt grafikte göster.
ax2.set_title("Kırpılmış Resim")
# Alt grafiğin başlığını ayarla.
ax2.axis("off")
# Eksenleri gizle.

plt.show()
# Oluşturulan tüm alt grafikleri içeren görselleştirme penceresini ekranda göster.
```
